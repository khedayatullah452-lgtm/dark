[ваш текст]


Данный проект представляет собой систему управления сотрудниками компании 
(Employee Management System), разработанную на языке C#. 

Цель проекта — продемонстрировать основные принципы объектно-ориентированного 
программирования: инкапсуляцию, наследование, полиморфизм и абстракцию.

Базовый класс Employee содержит:
- Атрибуты: Name (имя), Age (возраст), Salary (зарплата)
- Методы: GetInfo(), Work(), TakeVacation()

На основе базового класса созданы три производных класса:

1. Manager (Менеджер) — содержит дополнительный атрибут SubordinatesCount 
   (количество подчинённых). Метод Work() переопределён для демонстрации 
   управления командой.

2. Developer (Разработчик) — содержит дополнительный атрибут Specialization 
   (специализация). Метод TakeVacation() переопределён для согласования 
   отпуска с проектными сроками.

3. Designer (Дизайнер) — содержит дополнительный атрибут DesignStyle 
   (стиль дизайна). Метод Work() переопределён для работы над 
   дизайн-проектами.

Инкапсуляция реализована через приватные поля и свойства с проверкой 
корректности данных (возраст, зарплата).

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

// =====================================================
// Базовый класс: Employee
// =====================================================
public abstract class Employee
{
    private string _name;
    private int _age;
    private decimal _salary;

    public string Name
    {
        get => _name;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя не может быть пустым");
            _name = value;
        }
    }

    public int Age
    {
        get => _age;
        set
        {
            if (value < 18 || value > 70)
                throw new ArgumentException("Возраст должен быть от 18 до 70 лет");
            _age = value;
        }
    }

    public decimal Salary
    {
        get => _salary;
        set
        {
            if (value < 0)
                throw new ArgumentException("Зарплата не может быть отрицательной");
            _salary = value;
        }
    }

    protected Employee(string name, int age, decimal salary)
    {
        Name = name;
        Age = age;
        Salary = salary;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Зарплата: {Salary:N0} руб.";
    }

    public virtual void Work()
    {
        Console.WriteLine($"{Name} выполняет общие рабочие обязанности.");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"{Name} ушёл в отпуск.");
    }
}

// =====================================================
// Manager
// =====================================================
public class Manager : Employee
{
    public int SubordinatesCount { get; set; }

    public Manager(string name, int age, decimal salary, int subordinatesCount)
        : base(name, age, salary)
    {
        SubordinatesCount = subordinatesCount;
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} как менеджер руководит командой из {SubordinatesCount} человек.");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Подчинённых: {SubordinatesCount}";
    }
}

// =====================================================
// Developer
// =====================================================
public class Developer : Employee
{
    public string Specialization { get; set; }

    public Developer(string name, int age, decimal salary, string specialization)
        : base(name, age, salary)
    {
        Specialization = specialization;
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} пишет код в области {Specialization}.");
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"{Name} ({Specialization}) должен согласовать отпуск с проектными сроками.");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Специализация: {Specialization}";
    }
}

// =====================================================
// Designer
// =====================================================
public class Designer : Employee
{
    public string DesignStyle { get; set; }

    public Designer(string name, int age, decimal salary, string designStyle)
        : base(name, age, salary)
    {
        DesignStyle = designStyle;
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} работает над дизайн-проектом в стиле {DesignStyle}.");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Стиль дизайна: {DesignStyle}";
    }
}
```

```

// Демонстрация (без Main — напрямую)
// =====================================================
var employees = new List<Employee>
{
    new Manager("Иван Петров", 40, 500000m, 8),
    new Developer("Мария Сидорова", 28, 350000m, "C# / .NET"),
    new Designer("Алексей Иванов", 32, 300000m, "Минимализм")
};

Console.WriteLine("========== Информация о сотрудниках ==========\n");
foreach (var emp in employees)
{
    Console.WriteLine("---");
    Console.WriteLine(emp.GetInfo());
}

Console.WriteLine("\n========== Выполнение работы (Work) ==========\n");
foreach (var emp in employees)
{
    emp.Work();
}

Console.WriteLine("\n========== Отпуск (TakeVacation) ==========\n");
foreach (var emp in employees)
{
    emp.TakeVacation();
}

Console.WriteLine("\n========== Тест инкапсуляции ==========\n");
try
{
    var invalidEmp = new Developer("Тест", 15, 1000m, "Тест");
}
catch (ArgumentException ex)
{
    Console.WriteLine($"Ошибка при создании: {ex.Message}");
}

Console.WriteLine("\nПрограмма завершена.");
```
